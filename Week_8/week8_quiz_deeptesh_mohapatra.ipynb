# Week 8 Quiz - Single Agent Systems and Agent Pipelines

This quiz has 10 questions on how single agent systems and agent pipelines work - stateful
graphs, tools, routing, loops, error handling, evaluation and metrics. Each question below shows
the options, the correct answer, and a short explanation.

### Q1. What is a stateful directed graph in an agent pipeline?

A. A linear sequence of steps that cannot branch
B. A workflow where each step can store and use information from previous steps, with branching and looping
C. A database that stores agent passwords
D. A random collection of tools with no order

Correct answer: B

Explanation: Unlike a plain linear pipeline, a stateful directed graph carries a shared state
through the steps and can branch and loop based on that state.

### Q2. In an agent workflow graph, what do nodes and edges represent?

A. Nodes are data files and edges are models
B. Nodes represent tasks (for example a Calculator tool) and edges define how information flows between them
C. Nodes are users and edges are questions
D. Nodes and edges are the same thing

Correct answer: B

Explanation: A node is a task or tool; an edge is the connection that passes information and
control from one task to the next.

### Q3. What does conditional routing do in an agent pipeline?

A. It runs every tool for every query
B. It sends a query to the right handler based on its intent - calculations go to a calculator, keyword requests go to a keyword tool, and everything else goes to a general handler
C. It picks a tool at random
D. It stops the pipeline immediately

Correct answer: B

Explanation: Conditional routing chooses the next node based on the query's intent, so each
request is handled by the most suitable tool.

### Q4. Why are loops important in an agent graph?

A. They make the code shorter
B. They let the agent retry a step when it fails temporarily, which improves resilience instead of returning an error right away
C. They are required by the Python language
D. They exist only to slow the system down

Correct answer: B

Explanation: Loops enable retry mechanisms, so a temporary failure can be recovered from rather
than crashing the whole pipeline.

### Q5. How can a single agent simulate a multi-agent system?

A. By running on several computers at once
B. By dividing the work into different roles that each perform a distinct function, one after another
C. By being written in several programming languages
D. A single agent cannot simulate multiple agents

Correct answer: B

Explanation: One system can play several roles in sequence - for example a coordinator, a few
specialists, and a reviewer - which mimics a team of agents.

### Q6. What is the purpose of a JSON schema for tools?

A. To store the agent's long term memory
B. To define a standard structure for the data exchanged between agents and tools, including the required fields and their data types
C. To compress the tool output
D. To encrypt messages between tools

Correct answer: B

Explanation: A JSON schema describes each tool in a standard way, so the agent knows exactly what
inputs the tool needs and what it returns.

### Q7. When should tools be executed sequentially versus in parallel?

A. Always sequentially, no matter what
B. Sequentially when a task depends on the result of a previous task, and in parallel when tasks are independent, which is more efficient
C. Always in parallel
D. The order never matters

Correct answer: B

Explanation: Dependencies force a sequential order; independent tasks can run in parallel to save
time.

### Q8. Which of these are good error handling strategies in an agent pipeline?

A. Ignoring errors and hoping they go away
B. Using try-except blocks, retry mechanisms, and comprehensive logging
C. Restarting the computer on every error
D. Only printing the final answer and nothing else

Correct answer: B

Explanation: Catch errors with try-except, retry the ones that are temporary, and log what
happened so problems can be diagnosed.

### Q9. What does trajectory evaluation examine?

A. Only the final output of the agent
B. The complete sequence of decisions and intermediate steps the agent took, not just the final answer
C. The size of the language model
D. The number of tools that are installed

Correct answer: B

Explanation: Trajectory evaluation looks at the whole decision path, so you can tell whether the
agent reached the answer for the right reasons and took a sensible route.

### Q10. What do the task completion rate and cost metrics measure, and how are they used together?

A. Task completion rate measures internet speed and cost measures screen size
B. Task completion rate is the percentage of tasks completed successfully, cost metrics track the resources used (such as steps, tokens or time), and the two are balanced during optimization
C. They are two names for the same number
D. They only measure the accuracy of the language model

Correct answer: B

Explanation: You want a high task completion rate while keeping the cost reasonable, so tuning an
agent means balancing these two metrics.

## Summary

Answer key: 1-B, 2-B, 3-B, 4-B, 5-B, 6-B, 7-B, 8-B, 9-B, 10-B.

Key takeaways from the quiz:

- An agent pipeline is best modelled as a stateful directed graph of nodes (tasks) and edges (flow).
- Conditional routing sends each query to the right tool, and loops let it retry on temporary failures.
- Tools are described with JSON schemas so inputs and outputs are consistent.
- Good error handling (try-except, retries, logging) keeps the pipeline robust.
- Agents are judged on their whole trajectory, not just the final answer, and on metrics like task
  completion rate balanced against cost.
